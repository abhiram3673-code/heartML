```python
df_encode=pd.get_dummies(df,drop_first=True)
```

```python
df_encode
```

```python
df_encode.rename(columns={'HeartDisease':'is_HeartDisease','Sex_M':'is_Male'},inplace=True)
```

```python
df_encode=df_encode.astype(int)
```

```python
from sklearn.preprocessing import StandardScaler
cols=['Age', 'RestingBP', 'Cholesterol', 'MaxHR', 'Oldpeak']
scaler=StandardScaler()
df_encode[cols]=scaler.fit_transform(df_encode[cols])
```

Feature Engineering and Extraction

```python
from scipy.stats import pearsonr
selected_feat=['Age','RestingBP', 'Cholesterol','MaxHR', 'Oldpeak']
correlations={
    feature:pearsonr(df_encode[feature],df_encode['is_HeartDisease'])[0] for feature in selected_feat
}
corr_df=pd.DataFrame(list(correlations.items()),columns=['Feature','Correlation_Value'])
corr_df.sort_values(by='Correlation_Value',ascending=False)

```

```python
cat_feat=['FastingBS', 'is_Male', 'ChestPainType_ATA', 'ChestPainType_NAP',
       'ChestPainType_TA', 'RestingECG_Normal', 'RestingECG_ST',
       'ExerciseAngina_Y', 'ST_Slope_Flat', 'ST_Slope_Up']
from scipy.stats import chi2_contingency
alpha=0.05
chi2_results={}
for feat in cat_feat:
    cont=pd.crosstab(df_encode[feat],df_encode['is_HeartDisease'])
    chi2_stat,p_val,_,_=chi2_contingency(cont)
    des='Reject Null(Keep Feature)' if p_val<=alpha else 'Accept Null(Reject Feature)'
    chi2_results[feat]={
        'chi2_stats':chi2_stat,
        'p_value':p_val,
        'Decision':des
    }
chi2_df=pd.DataFrame(chi2_results).T
chi2_df=chi2_df.sort_values(by='p_value')
chi2_df
```
